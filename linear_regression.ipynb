```python
import numpy as np

class LinearRegression1:
    def __init__(self, learning_rate, epochs):
        self.learning_rate = learning_rate
        self.epochs = epochs

    def fit(self, X, Y):
        self.m, self.n = X.shape
        self.w = np.zeros(self.n)
        self.b = 0

        for _ in range(self.epochs):
            y_pred = np.dot(X, self.w) + self.b

            dw = (2 / self.m) * np.dot(X.T, (y_pred - Y))
            db = (2 / self.m) * np.sum(y_pred - Y)

            self.w -= self.learning_rate * dw
            self.b -= self.learning_rate * db

    def predict(self, X):
        v=np.dot(X, self.w) + self.b
        
        return float(v)



        

        
            
        
    
```

```python
import numpy as np

v = np.array([1,2,3,4,5,6]).reshape(-1,1)
u = np.array([100,400,700,1000,1300,1600])

model = LinearRegression1(0.01, 10000)
model.fit(v, u)

print(model.predict(np.array([[7]])))
```

```python
import numpy as np

class logistic_regression1:

    def __init__(self, learning_rate, epocs):
        self.learning_rate = learning_rate
        self.epocs = epocs

    # sigmoid function
    def c(self, a):
        return 1 / (1 + np.exp(-a))

    def fit(self, x, y):

        self.m, self.n = x.shape

        self.w = np.zeros(self.n)
        self.b = 0

        for i in range(self.epocs):

            # linear equation
            z = np.dot(x, self.w) + self.b

            # sigmoid output
            v = self.c(z)

            # gradients
            dw = (1 / self.m) * np.dot(x.T, (v - y))

            db = (1 / self.m) * np.sum(v - y)

            # update
            self.w = self.w - self.learning_rate * dw

            self.b = self.b - self.learning_rate * db

    def prediction(self, a):

        z = np.dot(a, self.w) + self.b

        v = self.c(z)

        return [1 if i >= 0.5 else 0 for i in v]
```
